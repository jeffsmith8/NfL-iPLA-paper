```python
# ── Environment setup ────────────────────────────────────────────────────────
# Adding the parent directory to sys.path allows `from ProxiPal import *`
# regardless of the notebook's working directory.

import sys
from pathlib import Path

module_dir = Path.cwd().parent
sys.path.append(str(module_dir))

from ProxiPal import *
# ProxiPal also imports pandas, numpy, scipy, matplotlib — these are
# available after this line without separate import statements.
# See: https://github.com/jeffsmith8/NfL-iPLA-paper

import re
from collections import OrderedDict
from matplotlib.ticker import LogLocator, FuncFormatter, NullFormatter
```

```python
stability_df = pd.read_csv('LiHep_CollectionStabilityRaw.csv')


stability_df_means = stability_df.drop_duplicates(subset=['sample_id', 'rep_id', 
                                                                    'SLR; log10(x); exc_std0; ct; mean_ng/L', 
                                                                    'filepath_csv'], keep='first')

def sort_key(series):
    # Extract components: group and timepoint
    return series.str.extract(r'^(.*?)(\s0hr|\s2hr|\s24hr)$') \
                 .apply(lambda row: (row[0], {' 0hr': 0, ' 2hr': 1, ' 24hr': 2}[row[1]]), axis=1)

sorted_df = stability_df_means[['sample_id', 'SLR; log10(x); exc_std0; ct; mean_ng/L', 'cv_slr_raw_ng/L']].sort_values(by='sample_id', key=sort_key)
sorted_df
```

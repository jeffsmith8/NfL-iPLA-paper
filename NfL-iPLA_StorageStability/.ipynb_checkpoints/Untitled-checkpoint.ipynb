### NfL Stability After Processing
For similar NfL reports: 2025 Lifeng, 2018 Keshavan, 2013 Kuhle; 2023 Rosadas

It is generally agreed that NfL from CSF, plasma, or serum is stable in storage and tolerant of multiple freeze-thaws (2025 Lifeng, 2018 Keshavan, 2023 Rosadas). As explained in the NfL collection stability report, nearly all such studies have relied on detection by the same anti-NfL clones. To this end we checked this is also true of NfL analysed by NfL-PLA which relies on a different antibody set. The results are good and relevant for future use with biobanked samples.

```python
storage_df = pd.read_csv("StorageStability.csv")
```

```python
view_cols = ['sample_id', 'rep_id', 'SLR; log10(x); exc_std0; ct; mean_ng/L']
storage_df[view_cols]
```

```python
storage_df_means = storage_df[view_cols].drop_duplicates()
storage_df_means
```

```python
# calculate recovery based on the 0x sample
QC_H_means = storage_df_means[storage_df_means['sample_id'].str.contains('NFL-QC-H')]
QC_M_means = storage_df_means[storage_df_means['sample_id'].str.contains('NFL-QC-M')]
QC_L_means = storage_df_means[storage_df_means['sample_id'].str.contains('NFL-QC-L')]
```

```python
# Ensure QC_H_means is an independent copy
QC_H_means = QC_H_means.copy()

# Get the zero concentration value
QC_H_zero = QC_H_means.loc[QC_H_means['sample_id'] == 'NFL-QC-H 0x', 'SLR; log10(x); exc_std0; ct; mean_ng/L'].values[0]

# Calculate recovery %
QC_H_means['recovery %'] = (100 * (QC_H_means['SLR; log10(x); exc_std0; ct; mean_ng/L'] / QC_H_zero)).round(2)
# Display
QC_H_means
```

```python
# Ensure QC_H_means is an independent copy
QC_M_means = QC_M_means.copy()

# Get the zero concentration value
QC_M_zero = QC_M_means.loc[QC_M_means['sample_id'] == 'NFL-QC-M 0x', 'SLR; log10(x); exc_std0; ct; mean_ng/L'].values[0]

# Calculate recovery %
QC_M_means['recovery %'] = (100 * (QC_M_means['SLR; log10(x); exc_std0; ct; mean_ng/L'] / QC_M_zero)).round(2)

# Display
QC_M_means
```

```python
# Ensure QC_H_means is an independent copy
QC_L_means = QC_L_means.copy()

# Get the zero concentration value
QC_L_zero = QC_L_means.loc[QC_L_means['sample_id'] == 'NFL-QC-L 0x', 'SLR; log10(x); exc_std0; ct; mean_ng/L'].values[0]

# Calculate recovery %
QC_L_means['recovery %'] = (100 * (QC_L_means['SLR; log10(x); exc_std0; ct; mean_ng/L'] / QC_L_zero)).round(2)

# Display
QC_L_means
```

```python
# Concatenate the 3 dataframes and manipulate in excel
# Add 'db' column to each before concatenation
QC_H_means['db'] = 'QC_H'
QC_M_means['db'] = 'QC_M'
QC_L_means['db'] = 'QC_L'

# Concatenate lengthwise
StorageStabilityTableData = pd.concat([QC_H_means, QC_M_means, QC_L_means], ignore_index=True)

# Save
StorageStabilityTableData.to_csv('StorageStabilityTableData.csv')

StorageStabilityTableData
```
